bin to ply

```python
import numpy as np
import open3d as o3d
from pathlib import Path
from hloc.utils.read_write_model import read_points3D_binary

def convert_bin_to_ply(bin_path: str, ply_path: str):
    bin_path = Path(bin_path)
    ply_path = Path(ply_path)

    print(f"Reading points from: {bin_path}")
    points3D = read_points3D_binary(str(bin_path))

    if not points3D:
        print("No 3D points found.")
        return

    xyz = np.array([p.xyz for p in points3D.values()])
    rgb = np.array([p.rgb for p in points3D.values()]) / 255.0

    pcd = o3d.geometry.PointCloud()
    pcd.points = o3d.utility.Vector3dVector(xyz)
    pcd.colors = o3d.utility.Vector3dVector(rgb)

    print(f"Saving to: {ply_path}")
    o3d.io.write_point_cloud(str(ply_path), pcd)
    print("Done!")

# 示例用法
if __name__ == "__main__":
    bin_file = 'outputs/sift+NN-ratio/points3D.bin'
    ply_file = 'sfm20.ply'
    convert_bin_to_ply(bin_file, ply_file)
```

bin to txt

```python
from hloc.utils.read_write_model import (
    read_model, write_model
)
import os

# 输入 bin 模型目录
model_path = 'outputs/sift+NN-ratio'
output_path = 'outputs/3txt'

os.makedirs(output_path, exist_ok=True)

# 读取 bin 格式模型
cameras, images, points3D = read_model(model_path, ext='.bin')

# 保存为 txt 格式
write_model(cameras, images, points3D, output_path, ext='.txt')
```
